```python
#2025-2의 정확한 합격자는 최종합격자가 아니라, 서류합격대상자입니다. 그래서 제외하고 분석하겠습니다.
#'커트라인 데이터 (6) (1).xlsx'은 마지막 2025-2 삭제한 엑셀 파일입니다.
#전체 데이터 규모: 총 952행 (2025-2 시트가 정상적으로 제외되었습니다.)
#저장된 파일명: total_data_v3.csv

import pandas as pd
# 1. 파일 읽기 및 시트 통합
file_path = '커트라인 데이터 (6) (1).xlsx'
xls = pd.ExcelFile(file_path)

all_sheets = []
for sheet_name in xls.sheet_names:
    # 데이터가 시작되는 두 번째 행(index 1)을 header로 지정
    df = pd.read_excel(file_path, sheet_name=sheet_name, header=1)

    # 컬럼 이름 표준화
    cols = ['구분', '자치구', '주택형', '주소지', '성별', '당첨_순위', '당첨_가점', '예비_순위', '예비_가점']
    df.columns = cols[:len(df.columns)]

    # 시트명을 공급시기로 추가
    df['공급시기'] = sheet_name
    all_sheets.append(df)

# 통합 데이터프레임 생성
total_df = pd.concat(all_sheets, ignore_index=True)

# 2. CSV 파일 저장
total_df.to_csv('total_data_v3.csv', index=False, encoding='utf-8-sig')

# 3. 데이터 확인
print(f"전체 행 개수: {len(total_df)}")
display(total_df.head())   # 상위 5개
display(total_df.tail())   # 하위 5개
display(total_df.sample(5)) # 무작위 5개
```

```python
#2021-1은 열에 '주택형'이 따로 있음. 나머지 데이터는 '주택명' 열에 대괄호[]안에 따로 적혀 있음.
# 기존 데이터 유지: '주택형' 열에 이미 숫자가 포함된 유효한 데이터가 있는 경우(예: 2021-1 시트 등)에는 해당 값을 그대로 유지합니다.
# 대괄호 추출: '주택형'이 비어있거나 -인 경우, 주소 정보에 포함된 [32A]와 같은 대괄호 안의 텍스트를 추출하여 '주택형_통합' 열에 저장합니다.
# 결과 저장: 보완된 데이터는 total_data_v4.csv 파일로 저장되었습니다.

import pandas as pd
import re

# 1. 이전 단계에서 생성된 CSV 로드
df = pd.read_csv('total_data_v3.csv')

def extract_housing_type(row):
    h_type = str(row['주택형']).strip()

    # 주택형 열에 유의미한 데이터(숫자 포함)가 있으면 유지
    if h_type != '-' and h_type.lower() != 'nan' and re.search(r'\d+', h_type):
        return h_type

    # 데이터가 없으면 주소지 열의 대괄호 [] 내용 추출
    address = str(row['주소지'])
    match = re.search(r'\[(.*?)\]', address)
    if match:
        return match.group(1)

    return h_type

# 통합 열 생성
df['주택형_통합'] = df.apply(extract_housing_type, axis=1)

# 확인용 출력
display_cols = ['공급시기', '자치구', '주택형', '주소지', '주택형_통합']
print("상위 데이터 (2021-1):\n", df[display_cols].head())
print("\n하위 데이터 (대괄호 추출 확인):\n", df[display_cols].tail())
print("\n무작위 샘플:\n", df[display_cols].sample(10))

# 파일 저장
df.to_csv('total_data_v4.csv', index=False, encoding='utf-8-sig')
```

```python
#복사본인 df_no_rank3에서 주소지는 동일하지만 주택형_통합이 서로 다른 데이터를 필터링하는 코드와 결과입니다.
# 이 과정을 통해 같은 건물 내에서 평형대에 따라 당첨 커트라인이 어떻게 차이 나는지 비교 분석할 수 있는 기반이 마련되었습니다.
#데이터 필터링 결과 요약
# 데이터 규모: 동일 주소 내에서 여러 평형이 존재하는 데이터는 총 505건이며, 이에 해당하는 주소지는 200곳입니다.
import pandas as pd

# 1. df_no_rank3 준비 (v4 로드 후 3순위 제외)
df = pd.read_csv('total_data_v4.csv')
df_no_rank3 = df[~df['당첨_순위'].astype(str).str.contains('3순위', na=False)].copy()

# 2. 동일 주소지 내에서 '주택형_통합'이 2개 이상인 주소지 추출
address_group = df_no_rank3.groupby('주소지')['주택형_통합'].nunique()
multi_type_addresses = address_group[address_group > 1].index

# 3. 해당 주소지 데이터 필터링 및 주소지별 정렬
df_multi_type = df_no_rank3[df_no_rank3['주소지'].isin(multi_type_addresses)].copy()
df_multi_type = df_multi_type.sort_values(by=['주소지', '주택형_통합'])

# 4. 결과 확인
print(f"🏠 동일 주소 내 다양한 주택형이 존재하는 데이터 수: {len(df_multi_type)}개")
print(f"📍 대상 주소지 수: {len(multi_type_addresses)}곳")
print("-" * 60)

# 주요 열 확인 (Head, Tail, Sample)
display_cols = ['공급시기', '자치구', '주소지', '주택형_통합', '당첨_순위', '당첨_가점']
display(df_multi_type[display_cols].head())
display(df_multi_type[display_cols].tail())
display(df_multi_type[display_cols].sample(5))
```

```python
#가설 검증을 위해 df_multi_type 데이터프레임에 순위와 가점을 결합한 새로운 기준인 '보정_커트라인' 열을 추가했습니다.
#순위 우선 원칙: 행복주택 당첨은 가점보다 순위가 우선하기 때문에, 1순위 데이터에 가중치(+14점)를 부여하여 2순위 데이터보다 항상 높은 점수를 갖도록 설계했습니다.
#계산 방식:
#1순위: 당첨 가점 + 14 (예: 1순위 7점 → 21점, 1순위 2점 → 16점)
#2순위: 당첨 가점 숫자 그대로 유지 (예: 2순위 6점 → 6점)
#데이터 정제: 가점 열과 순위 열의 숫자가 뒤바뀐 경우를 대비해 양쪽 열을 모두 검색하여 숫자를 추출하도록 코드를 보완했습니다.

import pandas as pd
import re

# 1. 보정_커트라인 계산 함수 정의
def calculate_adjusted_score(row):
    rank_str = str(row['당첨_순위'])
    score_val = str(row['당첨_가점'])

    # 숫자 추출 (가점 열 우선, 없으면 순위 열 검색)
    match = re.search(r'\d+', score_val)
    if not match: match = re.search(r'\d+', rank_str)
    score = int(match.group()) if match else 0

    # 1순위인 경우 +14 가중치 부여
    if '1순위' in rank_str or '1순위' in score_val:
        return score + 14
    return score

# 2. 새로운 열 추가
df_multi_type['보정_커트라인'] = df_multi_type.apply(calculate_adjusted_score, axis=1)

# 3. 데이터 확인 (Head, Tail, Sample)
display_cols = ['공급시기', '자치구', '주소지', '주택형_통합', '당첨_순위', '당첨_가점', '보정_커트라인']
print("[ 상위 5개 (Head) ]")
display(df_multi_type[display_cols].head())
print("\n[ 하위 5개 (Tail) ]")
display(df_multi_type[display_cols].tail())
print("\n[ 무작위 5개 (Sample) ]")
display(df_multi_type[display_cols].sample(5))
```

```python
#가설인 "평수가 넓어지면 커트라인도 높아진다"를 검증하기 위해 동일 주소지 내 평형별 데이터를 분석한 결과입니다.
#분석 대상 주소지: 194곳
#상관계수 (Pearson's r): 0.2664 (약한 양의 상관관계)
#통계적 유의성 (P-value): 1.8568e-09 (매우 유의미함)
#가설 일치율: 82.47% (분석 대상 주소지 194곳 중 160곳에서 평수 증가 시 커트라인 유지 혹은 상승)

#데이터 해석 및 결론
#가설은 통계적으로 지지됩니다: P-value가 기준치(0.05)보다 압도적으로 낮아, 평수와 커트라인 사이에는 우연이 아닌 확실한 연관성이 존재합니다.
#하지만 절대적인 것은 아닙니다: 상관계수(0.2664)가 아주 높지는 않은데, 이는 평수가 커져도 경쟁률이 낮거나 비인기 단지일 경우 점수가 크게 오르지 않는 예외 사례들이 존재하기 때문입니다.
#동일 건물 내에서는 가설이 매우 강하게 작용합니다: 일치율이 82%가 넘는다는 것은, 내가 지원하려는 건물 내에서 평수만 고민한다면 더 넓은 평수의 컷이 높을 확률이 매우 높다는 것을 시사합니다

import pandas as pd
import re
from scipy import stats

# 1. 평수(숫자) 추출 및 분석용 데이터 준비
df_analysis = df_multi_type.copy()
df_analysis['평수'] = df_analysis['주택형_통합'].apply(lambda x: int(re.search(r'\d+', str(x)).group()) if re.search(r'\d+', str(x)) else None)
df_analysis = df_analysis.dropna(subset=['평수', '보정_커트라인'])

# 2. 전체 상관계수 계산
corr, p_val = stats.pearsonr(df_analysis['평수'], df_analysis['보정_커트라인'])

# 3. 주소지별 평수 증가에 따른 점수 변화 경향성 확인
def check_trend(group):
    if len(group) < 2: return None
    group = group.sort_values('평수')
    return group['보정_커트라인'].is_monotonic_increasing

trend_results = df_analysis.groupby('주소지').apply(check_trend).dropna()

# 4. 결과 출력
print(f"상관계수: {corr:.4f}, P-value: {p_val:.4e}")
print(f"가설 일치 주소지 비율: {trend_results.mean()*100:.2f}%")

# 5. 사례별 상세 비교표 (피벗 테이블)
sample_addr = trend_results[trend_results == True].head(5).index
report = df_analysis[df_analysis['주소지'].isin(sample_addr)].pivot_table(
    index='주소지', columns='평수', values='보정_커트라인'
).fillna('-')
display(report)
```

```python
#df_no_rank3 데이터를 바탕으로 자치구별, 연도별(공급시기별) 당첨 가점의 상세 통계(평균, 중앙값, 최저점, 표준편차)를 산출했습니다.
#df_no_rank3를 기반으로 자치구별, 공급 시기(연도)별로 묶어 당첨 가점의 통계를 산출했습니다. 분석의 정확도를 위해 1순위 당첨자에게는 가점 +14점의 가중치를 부여한 '보정_가점'을 기준으로 통계치를 계산했습니다.

import pandas as pd
import re

# 1. 데이터 로드 및 3순위 제외
df_v4 = pd.read_csv('total_data_v4.csv')
df_no_rank3 = df_v4[~df_v4['당첨_순위'].astype(str).str.contains('3순위', na=False)].copy()

# 2. 보정_가점 계산 (1순위 가중치 반영 및 오기입 교정)
def calculate_score(r):
    # 가점과 순위 열에서 숫자 추출
    score_num = int(re.search(r'\d+', str(r['당첨_가점'])).group()) if re.search(r'\d+', str(r['당첨_가점'])) else 0
    rank_num = int(re.search(r'\d+', str(r['당첨_순위'])).group()) if re.search(r'\d+', str(r['당첨_순위'])) else 0

    # 1순위 여부 확인 (가중치 +14)
    is_first_rank = 14 if '1순위' in str(r['당첨_순위']) or '1순위' in str(r['당첨_가점']) else 0

    # 가점 열에 가점이 있고 순위 열에 순위가 있는 정상 케이스 위주로 합산
    return score_num + is_first_rank

df_no_rank3['보정_가점'] = df_no_rank3.apply(calculate_score, axis=1)

# 3. 자치구 및 공급시기(연도)별 통계 산출
stats_df = df_no_rank3.groupby(['자치구', '공급시기']).agg(
    평균=('보정_가점', 'mean'),
    중앙값=('보정_가점', 'median'),
    최저가점=('보정_가점', 'min'),
    표준편차=('보정_가점', 'std'),
    데이터수=('보정_가점', 'count')
).reset_index()

# 4. 결과 출력 및 저장
display(stats_df)
stats_df.to_csv('district_yearly_stats_final.csv', index=False, encoding='utf-8-sig')
```

```python
# 자치구를 묶기
#S급 (최상위)_강남, 서초, 송파, 마포_압도적 직주근접, 강남권 인프라, 홍대/합정 등 청년 문화권.

#A급 (우수)_성동, 용산, 광진, 영등포, 동작, 강동_성수/여의도 배후지, 한강변, 주요 지하철 노선(2, 5, 9호선) 요충지.

#B급 (일반)_중구, 서대문, 종로, 성북, 양천, 강서, 동대문_무난한 주거 환경, 대학가 인근, 업무지구 접근성 양호.

#C급 (외곽)_도봉, 노원, 강북, 중랑, 금천, 관악, 구로, 은평_상대적으로 공급 물량이 많고 컷이 낮게 형성되는 지역.

import pandas as pd

# 1. 등급 매핑 리스트 작성
tier_map = {
    'S급': ['강남구', '서초구', '송파구', '마포구'],
    'A급': ['성동구', '광진구', '영등포구', '동작구', '강동구'], # 용산구 제외
    'B급': ['중구', '서대문구', '종로구', '성북구', '양천구', '강서구', '동대문구'],
    'C급': ['도봉구', '노원구', '강북구', '중랑구', '금천구', '관악구', '구로구', '은평구']
}

# 역매핑 사전 생성
rev_tier_map = {dist: tier for tier, dists in tier_map.items() for dist in dists}

# 2. df_no_rank3에 '입지등급' 열 추가
df_no_rank3['입지등급'] = df_no_rank3['자치구'].map(rev_tier_map)

# 3. 등급별/연도별 통계 산출 (stats_tier)
stats_tier = df_no_rank3.groupby(['입지등급', '공급시기']).agg(
    평균=('보정_가점', 'mean'),
    중앙값=('보정_가점', 'median'),
    최저가점=('보정_가점', 'min'),
    데이터수=('보정_가점', 'count')
).reset_index()

# 등급 순서 정렬 (S->A->B->C)
stats_tier['입지등급'] = pd.Categorical(stats_tier['입지등급'], categories=['S급', 'A급', 'B급', 'C급'], ordered=True)
stats_tier = stats_tier.sort_values(['입지등급', '공급시기'])

print("--- [ 입지 등급별 연도 추세 요약 ] ---")
display(stats_tier)
```
